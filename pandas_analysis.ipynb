### Step 1: Import Pandas & Load Datasets

We will import the `pandas` library and load the sample datasets (`coffee.csv` and `sales_data.csv`) that we created for this project.

```python
import pandas as pd

# Load CSV dataset (Coffee Sales Warmup)
coffee_df = pd.read_csv("sample_data/coffee.csv")

# Load CSV dataset (Detailed Sales Data)
sales_df = pd.read_csv("sample_data/sales_data.csv")

print("Datasets loaded successfully!")
```

### Step 2: Data Inspection

We will inspect the datasets to understand their structure. This includes:
- Displaying the first few rows (`head`)
- Displaying the last few rows (`tail`)
- Checking the names and data types of all columns
- Viewing summary information (non-null counts, memory usage)

```python
# Inspect the top 5 rows of the sales dataset
print("--- FIRST 5 ROWS (HEAD) ---")
display(sales_df.head())

# Inspect the bottom 5 rows of the sales dataset
print("\n--- LAST 5 ROWS (TAIL) ---")
display(sales_df.tail())

# Display data types of each column
print("\n--- DATA TYPES ---")
print(sales_df.dtypes)

# Display general DataFrame info
print("\n--- DATAFRAME INFO ---")
sales_df.info()
```

### Step 3: Compute Summary Statistics

We will compute the key summary statistics:
- Numeric metrics: mean, median, min, max, count, standard deviation, and percentiles.
- Categorical metrics: unique counts, top occurring value, and its frequency.

```python
# Full descriptive statistics for numeric columns
print("--- NUMERIC DESCRIPTIVE STATISTICS ---")
display(sales_df.describe())

# Compute specific statistics manually to demonstrate individual pandas functions
print("\n--- MANUAL SUMMARY STATISTICS FOR 'Revenue' ---")
print(f"Total Rows (Count): {sales_df['Revenue'].count()}")
print(f"Mean Revenue:       ${sales_df['Revenue'].mean():.2f}")
print(f"Median Revenue:     ${sales_df['Revenue'].median():.2f}")
print(f"Minimum Revenue:    ${sales_df['Revenue'].min():.2f}")
print(f"Maximum Revenue:    ${sales_df['Revenue'].max():.2f}")
print(f"Std Dev of Revenue: ${sales_df['Revenue'].std():.2f}")

# Descriptive statistics for categorical columns
print("\n--- CATEGORICAL SUMMARY STATISTICS ---")
display(sales_df.describe(include=['object']))
```

### Step 4: Slicing, Column Selection, and Row Filtering

We will now perform data subsetting and filtering operations:
- Selecting specific columns of interest (e.g., `Date`, `Product`, `Revenue`).
- Slicing specific rows (e.g., rows index 5 to 10 inclusive).
- Filtering rows based on conditional expressions:
  - Filter for transactions where **Revenue is greater than $1,000**.
  - Filter for transactions in the **Electronics category** sold in the **East region**.

```python
# 1. Select specific columns
selected_cols = sales_df[['Date', 'Product', 'Revenue', 'Region']]
print("--- SELECTED COLUMNS ---")
display(selected_cols.head())

# 2. Slice rows by index (rows index 5 to 10 inclusive)
sliced_rows = sales_df.iloc[5:11]
print("\n--- SLICED ROWS (Index 5 to 10) ---")
display(sliced_rows)

# 3. Filter rows: Revenue > $1000
high_revenue_df = sales_df[sales_df['Revenue'] > 1000]
print("\n--- FILTERED ROWS: Revenue > $1000 ---")
display(high_revenue_df)

# 4. Filter rows: Category is 'Electronics' AND Region is 'East'
east_electronics_df = sales_df[(sales_df['Category'] == 'Electronics') & (sales_df['Region'] == 'East')]
print("\n--- FILTERED ROWS: East Electronics Sales ---")
display(east_electronics_df)
```

### Step 5: Save Filtered Results to CSV and Excel

Finally, we will save the filtered results (`high_revenue_df` and `east_electronics_df`) to new files.
- `filtered_high_revenue.csv`
- `filtered_east_electronics.xlsx` (Excel format using `openpyxl` engine)

```python
import os

# Create folder for output if it doesn't exist
os.makedirs("output", exist_ok=True)

# Export high revenue transactions to CSV
high_revenue_df.to_csv("output/filtered_high_revenue.csv", index=False)
print("Saved filtered CSV to 'output/filtered_high_revenue.csv'")

# Export East electronics sales to Excel
east_electronics_df.to_excel("output/filtered_east_electronics.xlsx", index=False, sheet_name="East Electronics")
print("Saved filtered Excel to 'output/filtered_east_electronics.xlsx'")
```
